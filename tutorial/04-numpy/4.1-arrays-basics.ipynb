### Why NumPy?

In Module 2 you stored numbers in Python **lists**. Lists are flexible (they can hold anything),
but that flexibility makes math on them slow and wordy. To double every number in a list you
need a loop or a list comprehension:

```python
doubled = [x * 2 for x in prices]
```

**NumPy** (Numerical Python) gives you the `ndarray`, an *array* that holds numbers of one
type in a single block of memory. Math on a whole array is written once and runs in fast C
code underneath:

```python
doubled = prices * 2
```

NumPy is the foundation that pandas, matplotlib and scikit-learn are all built on. Every pandas
column you used in Module 3 is a NumPy array under the hood, so learning the basics here makes
the later modules easier.

| Python list                     | NumPy array                   |
|---------------------------------|-------------------------------|
| Any mix of types                | One type for the whole array  |
| Math needs a loop               | Math works on the whole array |
| Slow for large data             | Fast, written in C            |
| 1-dimensional (nested for more) | Any number of dimensions      |

## 1. Importing NumPy

By convention NumPy is always imported as `np`. You will see this alias in every tutorial and every codebase.

````python
import numpy as np

print(np.__version__)
````

## 2. Creating an array from a list

`np.array()` turns a Python list into an array:

````python
prices = [10.5, 20.0, 15.25, 30.0]

arr = np.array(prices)
print(arr)
print(type(arr))
````

Now the whole-array math from the introduction actually works:

````python
print(arr * 2)       # double every price
print(arr + 1)       # add 1 to every price
print(arr > 15)      # compare every price
````

Compare with what the same operations do to a plain list (`*` repeats the list, `+ 1` is an error):

````python
print(prices * 2)

try:
    prices + 1
except TypeError as e:
    print("Error:", e)
````

## 3. Array attributes: `shape`, `ndim`, `size`, `dtype`

Every array carries a few pieces of information about itself:

- `ndim` — how many dimensions (1 for a row of numbers, 2 for a table, ...)
- `shape` — how many items along each dimension, as a tuple
- `size` — total number of items
- `dtype` — the data type shared by **all** items (e.g. `int64`, `float64`, `bool`)

````python
print("ndim :", arr.ndim)
print("shape:", arr.shape)
print("size :", arr.size)
print("dtype:", arr.dtype)
````

Because an array has a single `dtype`, NumPy picks the most general type when you mix values. Ints mixed with a float become floats:

````python
ints = np.array([1, 2, 3])
mixed = np.array([1, 2, 3.5])

print(ints, ints.dtype)
print(mixed, mixed.dtype)
````

You can also ask for a specific type with the `dtype=` argument:

````python
as_float = np.array([1, 2, 3], dtype=float)
print(as_float, as_float.dtype)
````

## 4. Two-dimensional arrays (tables)

A list of lists becomes a 2D array — think rows and columns, like a small spreadsheet. `shape` is then `(rows, columns)`:

````python
table = np.array(
    [
        [1, 2, 3],
        [4, 5, 6],
    ]
)
print(table)
print("ndim :", table.ndim)
print("shape:", table.shape)   # 2 rows, 3 columns
print("size :", table.size)
````

## 5. Handy array constructors

You often need an array filled with a pattern rather than typed-in values. NumPy has a function for each common case.

`np.zeros(shape)` and `np.ones(shape)` — arrays filled with 0 or 1 (useful as a starting point you fill in later):

````python
print(np.zeros(4))
print(np.ones((2, 3)))   # pass a tuple for 2D
````

`np.arange(start, stop, step)` — like Python's `range()`, but returns an array. `stop` is **not** included:

````python
print(np.arange(5))          # 0..4
print(np.arange(2, 10, 2))   # 2, 4, 6, 8
print(np.arange(0, 1, 0.25)) # works with floats too
````

`np.linspace(start, stop, num)` — `num` evenly spaced values, and here `stop` **is** included. Great for plotting axes:

````python
print(np.linspace(0, 1, 5))
````

`np.full(shape, value)` — fill with any value you choose:

````python
print(np.full(3, 7.5))
````

## 6. Random numbers

Random arrays are handy for practice data and for machine learning later. Use a `Generator`
with a fixed `seed` so you get the **same** random numbers every time you run the cell —
this makes your results reproducible.

````python
rng = np.random.default_rng(seed=42)

print(rng.integers(low=0, high=10, size=5))   # 5 random ints from 0 to 9
print(rng.random(3))                          # 3 random floats between 0 and 1
print(rng.normal(loc=0, scale=1, size=3))     # 3 values from a normal (bell curve) distribution
````
