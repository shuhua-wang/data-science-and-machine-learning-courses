## 1. Indexing and slicing a 1D array

1D arrays index and slice exactly like Python lists: positions start at `0`, negative numbers count from the end, and `start:stop` leaves out `stop`.

````python
scores = np.array([88, 92, 75, 64, 95, 81])

print(scores[0])      # first item
print(scores[-1])     # last item
print(scores[1:4])    # positions 1, 2, 3
print(scores[:3])     # first three
print(scores[::2])    # every second item
````

Unlike lists, you can also pass a **list of positions** to grab several items at once (called *fancy indexing*):

````python
print(scores[[0, 2, 5]])
````

Assigning to a slice changes the original array in place:

````python
scores[1:3] = 0
print(scores)
````

## 2. Indexing a 2D array: `[row, column]`

For a 2D array you give two positions separated by a comma: `arr[row, col]`.
Use `:` on its own to mean "everything along this axis".

````python
table = np.array(
    [
        [10, 11, 12, 13],
        [20, 21, 22, 23],
        [30, 31, 32, 33],
    ]
)
print(table)
````

````python
print(table[0, 0])    # row 0, column 0
print(table[1, 3])    # row 1, column 3
print(table[2, -1])   # row 2, last column
````

````python
print(table[0])       # whole first row
print(table[0, :])    # same thing, written out
print(table[:, 1])    # whole second column
````

````python
print(table[0:2, 1:3])   # rows 0-1, columns 1-2 -> a smaller 2D block
````

> **Tip:** this `[row, col]` idea is exactly what pandas `.iloc[row, col]` does in Module 3.
> NumPy is where that syntax comes from.

## 3. Boolean masks (filtering)

Comparing an array to a value gives an array of `True`/`False` — a **mask**. Putting the mask
inside `[]` keeps only the items where the mask is `True`. This is the same trick as boolean
indexing in pandas.

````python
temps = np.array([18.2, 25.6, 30.1, 22.8, 27.4, 15.9])

hot = temps > 25
print(hot)
print(temps[hot])
````

Usually the mask is written inline. Combine conditions with `&` (and), `|` (or) and `~` (not) — remember the parentheses:

````python
print(temps[temps > 25])
print(temps[(temps > 20) & (temps < 28)])
print(temps[(temps < 18) | (temps > 30)])
````

Masks also let you replace values that match a condition:

````python
capped = temps.copy()
capped[capped > 25] = 25
print(capped)
````

## 4. Views vs copies

A slice of a NumPy array is a **view** of the same data, not a copy. Changing the slice changes
the original. This is fast, but it surprises people. Use `.copy()` when you want an independent array.

````python
original = np.array([1, 2, 3, 4, 5])

view = original[:3]
view[0] = 100
print("original after changing the view:", original)

original = np.array([1, 2, 3, 4, 5])
copy = original[:3].copy()
copy[0] = 100
print("original after changing the copy:", original)
````

## 5. Reshaping

`reshape` rearranges the same items into a new shape. The total number of items must stay the same (12 items can be `(3, 4)`, `(4, 3)`, `(2, 6)`, ...):

````python
numbers = np.arange(12)
print(numbers)

print(numbers.reshape(3, 4))
print(numbers.reshape(4, 3))
````

Use `-1` for one dimension and NumPy works it out for you:

````python
print(numbers.reshape(2, -1))   # 2 rows, however many columns fit
````

`.flatten()` (or `.ravel()`) goes back to 1D, and `.T` transposes rows and columns:

````python
grid = numbers.reshape(3, 4)
print(grid)
print(grid.flatten())
print(grid.T)
print(grid.T.shape)
````

## 6. Joining arrays

`np.concatenate` joins arrays end to end. `np.vstack` stacks rows on top of each other, `np.hstack` puts them side by side:

````python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate([a, b]))
print(np.vstack([a, b]))
print(np.hstack([a, b]))
````
