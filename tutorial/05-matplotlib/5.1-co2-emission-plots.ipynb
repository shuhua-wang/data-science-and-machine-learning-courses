# Lesson 5.1 — CO2 Emission Plots

**Objectives**
- Explore different plots with `Pandas` and `matplotlib` library
- Explore CO2 emission data by visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_file_path = "../../data/owid-co2-data.csv"
co2_emission = pd.read_csv(data_file_path, sep=",")
co2_per_capita = co2_emission.loc[
    co2_emission["country"].isin(
        ["Canada", "United States", "China", "India", "United Kingdom"]
    ),
    ["country", "year", "co2_per_capita"],
]
co2_per_capita.head(7)
```

## 1. One line plot

Carbon dioxide (CO₂) emissions from burning fossil fuels and industrial processes. This includes emissions from transport, electricity
generation, and heating, but not land-use change.

### Canadian CO2 emission per capita

```python
canada_co2 = co2_per_capita.loc[
    co2_per_capita["country"] == "Canada", ["year", "co2_per_capita"]
]
canada_co2
```

```python
canada_co2.plot(x="year", y="co2_per_capita", kind="line")
plt.title("Canada CO2 emissions per capita")
plt.xlabel("year")
plt.ylabel("co2 emission in mt")
plt.grid(True)
plt.show()
```

### World Annual CO2 emission Plot

```python
world_annual_co2 = co2_emission.loc[
    co2_emission["country"].isin(
        ["World"]
    ),
    ["country", "year", "co2"],
]

world_annual_co2["co2"] = world_annual_co2["co2"] / 1000

world_annual_co2.head(7)
```

```python
world_annual_co2.plot(x="year", y="co2", kind="line")
plt.title("World annual co2 emission")
plt.xlabel("year")
plt.ylabel("co2 emission in billion tonnes")
plt.grid(True)
plt.show()
```

Use `matplotlib` library to plot instead of `pandas` plot function, you need to prepare the data by yourself with converting `pandas DataFrame` to Numpy Array.

```python
world_annual_co2.head(7)
```

```python
# convert pandas DataFrame to numpy array
co2_array = world_annual_co2.loc[:, ["year", "co2"]].to_numpy()
co2_array[:5]
```

```python
# Prepare x and y data
x = co2_array[:, 0]
y = co2_array[:, 1]
```

```python
fig, ax = plt.subplots()
ax.plot(x, y)

ax.set(xlabel='year', ylabel='co2 emission in billion tonnes',
       title='World annual co2 emission')
ax.grid()

fig.savefig("world_annual_co2_emission.png")
plt.show()
```

## 2. Multiple lines plot

Reshaped DataFrame organized by given index / column values.

```python
co2_reshape = (
    co2_per_capita.pivot(index="year", columns="country", values="co2_per_capita")
    .rename_axis(columns=None)
    .reset_index()
    .fillna(0.0)
)
co2_reshape
```

```python
co2_reshape.plot(
    x="year",
    y=["Canada", "China", "India", "United Kingdom", "United States"],
    kind="line",
)
plt.title("CO2 emissions per capita")
plt.xlabel("year")
plt.ylabel("co2 emission in mt")
plt.grid(True)
plt.show()
```
