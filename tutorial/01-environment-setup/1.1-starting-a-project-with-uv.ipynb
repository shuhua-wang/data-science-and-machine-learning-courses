# Lesson 1.1 — Starting a Project with `uv`

## Step 1: Confirm `uv` is installed (in your terminal)

Run this in your terminal — not in this notebook, since it's a shell command, not
Python code:

```bash
uv --version
```

Expected output (your exact version/platform tag may differ slightly):

```text
uv 0.12.13 (0ebbd9274 2026-09-10 aarch64-apple-darwin)
```

## Step 2: Install / Manage Python using `uv`

Run this in your terminal

```bash
uv python list
```

Expected output (you exact version/platform may differ slightly):

```text
cpython-3.13.2-macos-aarch64-none   <download available>
cpython-3.10.12-macos-aarch64-none  /Users/yourname/.local/bin/python3.10
```

Install expected Python

```bash
uv python install 3.14
```

Expected output:

```text
Installed Python 3.14.7 in 1.18s
 + cpython-3.14.7-macos-aarch64-none (python3.14)
```

## Step 3: Manage a Project using `uv`

Open an empty folder, and initialize a project:

```bash
cd my_project
uv init  # initialize a project using uv
```

Expected output:

```text
Initialized project `my-project`
```

The project has following file structures:

```text
.
├── README.md
├── pyproject.toml
└── src
    └── my_project
        └── __init__.py
```

Pin a Python version into this new project:

```bash
uv python pin 3.14
```

Expected output:

```text
Pinned `.python-version` to `3.14`
```

Check if the python version is what you pin in the project:

```bash
uv run python --version
```

Expected output:

```text
Creating virtual environment at: .venv
      Built my-project @ file:///Users/shuhuawang/Desktop/my_project                                                                                                                     
Installed 1 package in 3ms
Python 3.14.7
```


## Step 4: Managing Python Packages

A general `pyproject.toml` file has following configurations, when you add or remove a Python packages,
its depdendencies configuration is updated.

```toml
[project]
name = "data-science-and-machine-learning-courses"
version = "0.1.0"
description = "Data Science and Machine Learning Courses Material"
readme = "README.md"
requires-python = ">=3.14"
dependencies = [
    "jupyter>=1.1.1",
    "matplotlib>=3.11.1",
    "notebook>=7.6.2",
    "numpy>=2.5.3",
    "openpyxl>=3.1.5",
    "scikit-learn>=1.9.0",
    "scipy>=1.18.1",
]
```

Add `pandas` into your project:

```bash
uv add pandas
```

Expected output:

```text
Resolved 117 packages in 429ms
Prepared 1 package in 730ms
Installed 1 package in 13ms
 + pandas==3.0.6
```

Remove `pandas` from your project:

```bash
uv remove pandas
```

Expected output:

```text
Resolved 116 packages in 756ms
Uninstalled 1 package in 210ms
 - pandas==3.0.5
```

Lock your dependencies into your project (`uv.lock` file), or update the project's lockfile

```bash
uv lock
```

Update the project's environment (run it when copy a project to a new environment):

```bash
uv sync
```
