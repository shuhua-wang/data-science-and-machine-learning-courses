# Lesson 1.1 — Starting a Project with `uv`

## Step 1: Confirm `uv` is installed (in your terminal)

Run this in your terminal — not in this notebook, since it's a shell command, not
Python code:

```bash
uv --version
```

Expected output (your exact version/platform tag may differ slightly):

```text
uv 0.12.13 (0ebbd9274 2026-09-10 aarch64-apple-darwin)
```

## Step 2: Install / Manage Python using `uv`

Run this in your terminal

```bash
uv python list
```

Expected output (you exact version/platform may differ slightly):

```text
cpython-3.13.2-macos-aarch64-none   <download available>
cpython-3.10.12-macos-aarch64-none  /Users/yourname/.local/bin/python3.10
```

Install expected Python

```bash
uv python install 3.14
```

Expected output:

```text
Installed Python 3.14.7 in 1.18s
 + cpython-3.14.7-macos-aarch64-none (python3.14)
```

## Step 3: Manage a Project using `uv`

Open an empty folder, and initialize a project:

```bash
cd my_project
uv init  # initialize a project using uv
```

Expected output:

```text
Initialized project `my-project`
```

The project has following file structures:

```text
.
├── README.md
├── pyproject.toml
└── src
    └── my_project
        └── __init__.py
```

Pin a Python version into this new project:

```bash
uv python pin 3.14
```

Expected output:

```text
Pinned `.python-version` to `3.14`
```

Check if the python version is what you pin in the project:

```bash
uv run python --version
```

Expected output:

```text
Creating virtual environment at: .venv
      Built my-project @ file:///Users/shuhuawang/Desktop/my_project                                                                                                                     
Installed 1 package in 3ms
Python 3.14.7
```
