## A `Series`: one labeled column

```python
import pandas as pd

ages = pd.Series([34, 28, 45, 31], name="age")
ages
```

```python
ages = pd.Series([34, 28, 45, 31], index=["c1", "c2", "c3", "c4"], name="age")
ages["c3"]
```

## A `DataFrame` from a dict of lists

```python
sample = pd.DataFrame({
    "customer_id": ["C00001", "C00002", "C00003"],
    "region": ["North", "South", "West"],
    "age": [34, 28, 45],
})
sample
```

```python
rows = [
    {"customer_id": "C00001", "region": "North", "age": 34},
    {"customer_id": "C00002", "region": "South", "age": 28},
]
pd.DataFrame(rows)
```

```python
type(sample["age"])  # picking a column out of a DataFrame gives a Series
```

## The index

```python
by_id = customers.set_index("customer_id")
by_id.loc["C00003"]
```

`set_index()` returns a **new** DataFrame — `customers` itself is untouched. Check:

```python
print("customer_id" in customers.columns)  # True: original DataFrame is unchanged
print("customer_id" in by_id.columns)        # False: it became the index instead
```
