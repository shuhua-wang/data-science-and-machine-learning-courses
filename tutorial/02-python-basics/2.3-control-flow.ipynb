**Control flow** is how a program decides which code to run, and how many times to run it, instead of just executing every line top to bottom. This lesson covers Python's two tools for that: conditionals (`if`) and loops (`for / while`).

## 1. `if` / `elif` / `else`

An `if` statement runs a block of code only when a condition is `True`

Two things to notice, because both are essential Python syntax:

- The line starts with `if`, followed by a condition, followed by a colon `:`
- The line(s) to run are indented — conventionally by 4 spaces.
- Python uses indentation (whitespace at the start of a line) to mark which lines belong to which block of code, instead of curly braces {} like many other languages. This is not a style choice you can skip — it's part of Python's syntax.
- Add `elif` (short for "else if") for additional conditions, and `else` to catch everything that didn't match

Python checks each condition top to bottom and runs the first block whose condition is `True`, then skips the rest — even if a later condition would also have been `True`. You can chain as many `elifs` as you need, and `else` (with no condition) is optional.

```python
temperature = 85

if temperature > 90:
    print("It's hot.")
elif temperature > 70:
    print("It's warm.")
else:
    print("It's cool.")
```

## 2. `for` loops

A loop repeats a block of code multiple times. A `for` loop repeats once for each item in a collection (like a list of names, or a range of numbers) — use it when you know (or can get) the set of things you want to go through.

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

To loop a specific number of times, use `range()`:

```python
for i in range(5):
    print(i)
```

`range(5)` produces the numbers `0, 1, 2, 3, 4` — five numbers starting at `0`, and stopping before `5`.

## 3. `while` loops

A while loop repeats as long as a condition stays `True` — use it when you don't know in advance how many times you'll need to loop, it only stops when the condition is met.

```python
count = 0

while count < 3:
    print(count)
    count += 1
```

Each time through the loop, Python checks the condition (`count < 3`) before running the block. The moment it's `False`, the loop stops.

## 4. `break` and `continue`

`break` exits a loop immediately, skipping any remaining iterations:

```python
numbers = [4, 8, 15, 16, 23, 42]

for number in numbers:
    if number == 16:
        print("Found 16, stopping.")
        break
    print(number)
```

`continue` skips the rest of the current iteration and jumps straight to the `next one` (`it does not exit the loop`):

```python
for number in range(1, 6):
    if number % 2 == 0:
        continue  # skip even numbers
    print(number)
```

## 5. Common usage: counting and accumulating

Two loop patterns come up constantly, so it's worth naming them explicitly.

`Counting` — keep a running and count how many times something happens:

```python
words = ["cat", "dog", "cat", "bird", "cat"]
cat_count = 0

for word in words:
    if word == "cat":
        cat_count += 1

print(cat_count)
```

`Accumulating` — build up a total (or a combined result) value across a loop:

```python
prices = [19.99, 5.50, 12.25]
total = 0

for price in prices:
    total += price

print(total)
```

Both patterns follow the same shape: create a starting variable before the loop (`cat_count = 0`, `total = 0`), then update it inside the loop.
