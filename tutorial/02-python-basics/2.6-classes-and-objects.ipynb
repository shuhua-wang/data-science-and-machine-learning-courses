## Defining a class

```python
class Dog:
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"

    def birthday(self):
        self.age += 1
        return f"{self.name} is now {self.age} years old."
```

- `__init__` runs automatically when a new `Dog` is created, and sets up its starting **attributes** (`self.name`, `self.breed`, `self.age`).
- `self` always refers to the specific object a method was called on.
- `bark` and `birthday` are **methods** - functions that belong to the class and can read or change that object's own data.

## Creating and using objects

```python
my_dog = Dog("Rex", "Labrador", 3)
print(my_dog.name)
print(my_dog.breed)
print(my_dog.bark())
```

```python
dog_a = Dog("Rex", "Labrador", 3)
dog_b = Dog("Biscuit", "Poodle", 5)

print(dog_a.bark())
print(dog_b.bark())
```

Each `Dog` object is independent, with its own attributes.

```python
print(dog_a.age)
print(dog_a.birthday())   # calling a method that changes the object's own state
print(dog_a.age)

print(dog_b.age)          # dog_b is untouched - each object's state is separate
```

> **Beginner mistake: forgetting `self`.** Every method needs `self` as its first parameter, even though you never pass it explicitly when calling the method - Python supplies it automatically. Forgetting it causes a `TypeError` the moment you call the method, as shown below.

```python
class BrokenDog:
    def __init__(self, name):
        self.name = name

    def bark():           # missing self!
        return "Woof!"

broken_dog = BrokenDog("Buddy")

try:
    broken_dog.bark()
except TypeError as error:
    print(f"Caught an error: {error}")
```

### Solution

```python
# Exercises 1-3 combined
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def summary(self):
        return f"'{self.title}' by {self.author} ({self.pages} pages)"

    def is_long(self):
        return self.pages > 400

book_a = Book("Dune", "Frank Herbert", 412)
book_b = Book("The Little Prince", "Antoine de Saint-Exupery", 96)

print(book_a.summary())
print(book_b.summary())

print(book_a.is_long())
print(book_b.is_long())
```
