## Lists

```python
groceries = ["eggs", "milk", "bread"]
print(groceries)
```

```python
print(groceries[0])   # first item
print(groceries[1])
print(groceries[-1])  # last item
```

> **Beginner mistake: off-by-one errors with indexing.** Python indexes start at `0`. A 3-item list has valid indexes `0`, `1`, `2` - `groceries[3]` raises an `IndexError`.

```python
try:
    print(groceries[3])
except IndexError as error:
    print(f"Caught an error: {error}")
```

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:3])   # index 1 up to, not including, index 3
print(numbers[:2])    # from the start
print(numbers[2:])    # to the end
```

```python
groceries.append("butter")        # add to the end
print(groceries)

groceries[0] = "brown eggs"       # change an item by index
print(groceries)

groceries.remove("bread")         # remove a specific value
print(groceries)

print(len(groceries))
```

## Tuples

```python
point = (3, 7)
print(point[0])
```

```python
try:
    point[0] = 9
except TypeError as error:
    print(f"Caught an error: {error}")
```

Tuples are ordered and indexable, just like lists - but **immutable**, so they're a good fit for fixed groups of values like coordinates.

## Dictionaries

```python
person = {"name": "Grace", "age": 36, "city": "Boston"}
print(person["name"])
```

```python
person["age"] = 37                # update a value
person["job"] = "Engineer"        # add a new key/value pair
print(person)
```

```python
print(person.get("job"))              # key exists
print(person.get("salary"))           # key missing -> None
print(person.get("salary", 0))        # key missing -> custom default
```

```python
for key, value in person.items():
    print(f"{key}: {value}")
```

## Sets

```python
tags = {"python", "beginner", "python", "tutorial"}
print(tags)   # the duplicate 'python' was dropped
```

```python
unique_visitors = set(["alice", "bob", "alice", "carol", "bob"])
print(unique_visitors)

print("alice" in unique_visitors)
print("dave" in unique_visitors)
```

## Which structure should I use?

| Structure | Ordered? | Changeable? | Duplicates? | Use it for |
|-----------|----------|-------------|-------------|------------|
| `list` | yes | yes | yes | a general-purpose sequence you'll modify |
| `tuple` | yes | no | yes | a fixed, small group of values |
| `dict` | yes (insertion order) | yes | keys must be unique | looking things up by name/key |
| `set` | no | yes | no (auto-removed) | uniqueness checks, membership tests |
