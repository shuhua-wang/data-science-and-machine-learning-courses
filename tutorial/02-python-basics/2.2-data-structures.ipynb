## 1. Lists

- A list is an ordered, changeable ("mutable") collection of items, written with square brackets `[]` and commas between items
- A list can hold any type, and even a mix of types, though in practice most lists hold one consistent type of thing
- Indexing - get a single item by its position, starting from `0`

```python
groceries = ["eggs", "milk", "bread"]
print(groceries)
```

```python
mix_types_list = ["apple", 3, True]
print(mix_types_list)
```

```python
print(groceries[0])   # first item
print(groceries[1])   # second item
print(groceries[2])   # third item
print(groceries[-1])  # last item
```

> **mistake: off-by-one errors with indexing.** Python indexes start at `0`. A 3-item list has valid indexes `0`, `1`, `2` - `groceries[3]` raises an `IndexError`.

```python
try:
    print(groceries[3])
except IndexError as error:
    print(f"Caught an error: {error}")
```

Slicing — get a sub-list using `start:stop` (stop is excluded):

```python
numbers = [10, 20, 30, 40, 50]
print(numbers[1:3])   # index 1 up to, not including, index 3. (1,2)
print(numbers[:2])    # from the start, (0,1)
print(numbers[2:])    # to the end, (2,3,4)
```

Mutating a list - lists can be changed in place after creation

Common list methods:

- `.append(x)`: adds `x` to the end
- `.remove(x)`: removes the first match of `x`
- `.sort(x)`: sorts the list in place
- `.pop(x)`: removes and returns the last item
- `len(some_list)`: tells you how many items are in the list

```python
groceries.append("butter")        # add to the end
print(groceries)

groceries[0] = "brown eggs"       # change an item by index
print(groceries)

groceries.remove("bread")         # remove a specific value
print(groceries)

print(len(groceries))
```

## 2. Tuples

- A tuples is an ordered, immutable collection of items, written with brackets `()` and commas between items

- A tuple looks and behaves almost exactly like a list — ordered, indexable, sliceable — except it's immutable: once created, it cannot be changed.

- Tuples are ordered and indexable, just like lists - but **immutable**, so they're a good fit for fixed groups of values like coordinates.

- A tuples can hold any type, and even a mix of types

```python
point = (3, 7)
print(point[0])
```

```python
point[0] = 9
```

Why use a tuple instead of a list?

Reach for a tuple when the collection represents a fixed, small group of values that shouldn't change — like an (x, y) coordinate, or a (latitude, longitude) pair. The immutability is a feature: it signals to anyone reading your code "this won't be modified," and Python will enforce that for you.

```python
mix_type_tuple = (1, "apple", True)
```

```python
print(mix_type_tuple[0])
print(mix_type_tuple[-1])
print(mix_type_tuple[1:])
```

## 3. Dictionaries

A dictionary (dict) stores key/value pairs — instead of looking items up by position (like a list), you look them up by a meaningful key. Written with curly braces `{}`:

```python
person = {"name": "Dave", "age": 36, "city": "Calgary"}
print(person["name"])
```

```python
person["age"] = 37                # update a value
person["job"] = "Engineer"        # add a new key/value pair
print(person)
```

```python
print(person["job"])              # key exists
```

```python
print(person["salary"])            # key not exists, raise an error
```

Looking up a key that doesn't exist raises a `KeyError`,
To look up safely, with a fallback value of the key is missing, use `.get()`

```python
print(person.get("job"))              # key exists
print(person.get("salary"))           # key missing -> None
print(person.get("salary", 0))        # key missing -> custom default
```

You can loop over a dictionary's keys, values, or both:

```python
for key, value in person.items():
    print(f"{key}: {value}")
```

`.item()` gives you both the key and value together on each pass - this is the pattern you'll use most often.

## 4. Sets

- A set is an unordered collection of unique items
- Duplicates are automatically removed, and there's no indexing (since there's no order to index into)
- Written with curly braces like a dict, but with values only, no key/value pairs:

```python
tags = {"python", "beginner", "python", "tutorial"}
print(tags)   # the duplicate 'python' was dropped
```

Sets exist for two main jobs:

- removing duplicates
- fast membership testing (checking whether something is in the collection):

```python
unique_visitors = set(["alice", "bob", "alice", "carol", "bob"])  # list is converted to a set using set()
print(unique_visitors)

print("alice" in unique_visitors)
print("dave" in unique_visitors)
```

## Which structure should I use?

| Structure | Ordered?              | Changeable? | Duplicates?         | Use it for                               |
|-----------|-----------------------|-------------|---------------------|------------------------------------------|
| `list`    | yes                   | yes         | yes                 | a general-purpose sequence you'll modify |
| `tuple`   | yes                   | no          | yes                 | a fixed, small group of values           |
| `dict`    | yes (insertion order) | yes         | keys must be unique | looking things up by name/key            |
| `set`     | no                    | yes         | no (auto-removed)   | uniqueness checks, membership tests      |
