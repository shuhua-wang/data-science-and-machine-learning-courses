## Creating and reassigning variables

```python
age = 29
print(age)
```

```python
# Reassignment: age now points to a new value
age = 30
print(age)
```

```python
# Using the current value to compute the new value
age = age + 1
print(age)
```

```python
# Shorthand for the same thing
age += 2
print(age)
```
