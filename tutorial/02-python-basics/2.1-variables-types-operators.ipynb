## Creating and reassigning variables

```python
age = 29
print(age)
```

```python
# Reassignment: age now points to a new value
age = 30
print(age)

# Using the current value to compute the new one
age = age + 1
print(age)

# Shorthand for the same thing
age += 1
print(age)
```

Notice each `print()` shows a different value - `age` is just a name pointing at whatever value it was last assigned.

## Python's core types

```python
print(type(29))
print(type(3.14))
print(type("hello"))
print(type(True))
print(type(None))
```

Five core types: `int` (whole numbers), `float` (decimals), `str` (text), `bool` (`True`/`False`), and `None` (no value at all).

## Strings and f-strings

```python
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name
print(full_name)
```

```python
# An f-string lets you drop variables directly into text
age = 36
message = f"{first_name} is {age} years old."
print(message)
```

## Dynamic typing

```python
data = 42
print(type(data))

data = "now I'm text"
print(type(data))
```

Python never needs a type declared up front - it figures the type out from the value you assign, and the same name can point to a different type later.

```python
# Converting between types explicitly
text_number = "42"
real_number = int(text_number)
print(real_number + 8)

print(int("42"))
print(float("3.14"))
print(str(42))
```

> **Beginner mistake: mixing strings and numbers.** `"42" + 8` raises a `TypeError` - Python won't silently guess you meant to add them. The cell below shows the error being caught, rather than crashing the whole notebook.

```python
try:
    result = "42" + 8
except TypeError as error:
    print(f"Caught an error: {error}")
```

## Operators

```python
print(7 + 3)  # addition
print(7 - 3)  # subtraction
print(7 * 3)  # multiplication
print(7 / 3)  # division -> always a float
print(7 // 3)  # floor division -> rounded down
print(7 % 3)  # modulo -> the remainder
print(7**2)  # exponent
```

```python
# floor division + modulo together: turn 130 minutes into hours and minutes
total_minutes = 130
hours = total_minutes // 60
minutes = total_minutes % 60
print(f"{hours}h {minutes}m")
```

```python
print(7 > 3)
print(7 < 3)
print(7 >= 7)
print(7 <= 3)
print(7 == 7)
print(7 != 3)
```

```python
age = 25
has_ticket = True

print(age >= 18 and has_ticket)  # both must be True
print(age >= 65 or has_ticket)  # at least one must be True
print(not has_ticket)  # flips True <-> False
```

> **Beginner mistake: off-by-one confusion with `>=` vs `>`.** "At least 18" means `age >= 18`, not `age > 18`, which would wrongly exclude someone who is exactly 18.
