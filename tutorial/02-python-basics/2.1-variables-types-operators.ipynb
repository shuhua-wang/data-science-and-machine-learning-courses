## 1. Creating and reassigning variables

Variable:

- A **name** that points to a value in memory.
- `=` assign a value to a variable (`==` is an operator used to compared two variables)
- Python variables are mutable (can be reassign)

```python
age = 29
print(age)
```

```python
# Reassignment: age now points to a new value
age = 30
print(age)

# Using the current value to compute the new one
age = age + 1
print(age)

# Shorthand for the same thing
age += 1
print(age)
```

Notice each `print()` shows a different value - `age` is just a name pointing at whatever value it was last assigned.

## 3. Python's core types

| Type    | Meaning         | Example          |
|---------|-----------------|------------------|
| `int`   | whole number    | `30`             |
| `float` | decimal number  | `3.14`           |
| `str`   | text            | `"hello"`        |
| `bool`  | truth value     | `True` / `False` |
| `None`  | no value at all | `None`           |

```python
print(type(30))
print(type(3.14))
print(type("hello"))
print(type(True))
print(type(None))
```

```python
pi = 3.14
print(pi)
print(type(pi))
```

```python
first_name = "Dave"
print(first_name)
print(type(first_name))
```

```python
is_sunday = True
print(is_sunday)
print(type(is_sunday))
```

```python
value = None
print(value)
print(type(value))
```

Five core types: `int` (whole numbers), `float` (decimals), `str` (text), `bool` (`True`/`False`), and `None` (no value at all).

## 4. Strings and f-strings

```python
first_name = "Ada"
last_name = "Lovelace"
full_name = first_name + " " + last_name
print(full_name)
```

f-strings:

- `f"..."` - put variables right inside `{}`
- standard, readable way to combine text + values

```python
# An f-string lets you drop variables directly into text
first_name = "Ada"
message = f"my first name is {first_name}."
print(message)
```

## 5. Dynamic typing

- Python is a dynamically typed language, meaning that variable types are determined at runtime, Python figures out the type from the value - no declarations needed.
- Python never needs a type declared up front - it figures the type out from the value you assign, and the same name can point to a different type later.

```python
data = 42
print(type(data))

data = "now I'm text"
print(type(data))
```

```python
# you also can add type annotation when defining the variable using a colon
# this is not mandatory, optional for better understanding

pi: float = 3.14
print(type(pi))
```

convert data type using `int()`, `float()`, `str()`, `bool()`

```python
# Converting between types explicitly
text_number = "42"
real_number = int(text_number) # convert a string to an integer
print(real_number + 8)
```

```python
# convert float to int
print(int(3.14))

# convert string to int
print(int("3"))

# convert int to float
print(float(5))

# convert int to bool
print(bool(1))
print(bool(0))  # 0 is converted to False
print(bool("name"))
print(bool("3.14"))
```

## 6. Operators

```python
print(f"addition: 7 + 3 = {7 + 3}")  # addition
print(f"substraction: 7 - 3 = {7 - 3}")  # subtraction
print(f"multiplication: 7 * 3 = {7 * 3}")  # multiplication
print(f"division: 7 / 3 = {7 / 3}")  # division -> always a float
print(f"floor division: 7 // 3 = {7 // 3}")  # floor division -> rounded down
print(f"modulo: 7 % 3 = {7 % 3}")  # modulo -> the remainder after dividing 7 by 3
print(f"exponent: 7**2 = {7**2}")  # exponent
```

```python
# Example: floor division + modulo together: turn 130 minutes into hours and minutes
total_minutes = 130
hours = total_minutes // 60
minutes = total_minutes % 60
print(f"{hours}h {minutes}m")
```

Comparison operators: always return a boolean

```python
print(7 > 3)
print(7 < 3)
print(7 >= 7)
print(7 <= 3)
print(7 == 7)
print(7 != 3)
```

`and`, `or`, and `not` are logical operators used to evaluate conditions and control the flow of your program

```python
age = 25
has_ticket = True

print(age >= 18 and has_ticket)  # both must be True
print(age >= 65 or has_ticket)  # at least one must be True
print(not has_ticket)  # flips True <-> False
```
