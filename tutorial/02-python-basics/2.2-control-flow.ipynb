## `if` / `elif` / `else`

```python
temperature = 85

if temperature > 90:
    print("It's scorching.")
elif temperature > 70:
    print("It's warm.")
else:
    print("It's cool.")
```

Only the **first** matching branch runs. Try changing `temperature` above to `95` or `50` and re-running the cell to see the other branches fire.

```python
age = 25
has_ticket = True

if age >= 18 and has_ticket:
    print("Entry allowed.")
else:
    print("Entry denied.")
```

## `for` loops

```python
fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

```python
for i in range(5):
    print(i)
```

> **Beginner mistake: off-by-one errors with `range()`.** `range(5)` gives `0` through `4` - five numbers, but the last one is `4`, not `5`. For `1` through `5` inclusive, use `range(1, 6)`.

## `while` loops

```python
count = 0

while count < 3:
    print(count)
    count += 1
```

> **Beginner mistake: the infinite loop.** If you forget to update the condition variable (here, `count += 1`), the loop never stops. If a notebook cell seems frozen, this is the first thing to check.

## `break` and `continue`

```python
numbers = [4, 8, 15, 16, 23, 42]

for number in numbers:
    if number == 16:
        print("Found 16, stopping.")
        break
    print(number)
```

```python
for number in range(1, 6):
    if number % 2 == 0:
        continue  # skip even numbers
    print(number)
```

## Common patterns: counting and accumulating

```python
words = ["cat", "dog", "cat", "bird", "cat"]
cat_count = 0

for word in words:
    if word == "cat":
        cat_count += 1

print(cat_count)
```

```python
prices = [19.99, 5.50, 12.25]
total = 0

for price in prices:
    total += price

print(total)
```
